```python
import pandas as pd
```

```python
df = pd.read_csv("data/processed/accidents_cleaned.csv")
```

```python
df_clean = df.dropna()
```

```python
print(df_clean.dtypes)
```

```python
df = df.replace(r'\xa0', '', regex=True)  
df = df.apply(pd.to_numeric, errors='coerce') 
```

```python
df.fillna(df.median(), inplace=True) 
```

```python
df_encoded = pd.get_dummies(df_clean, columns=["dep", "com", "adr", "actp"], drop_first=True)
```

```python
for col in ["dep", "com", "adr", "actp"]:
    top_categories = df_clean[col].value_counts().index[:50]  # Garder les 50 plus fréquentes
    df_clean[col] = df_clean[col].apply(lambda x: x if x in top_categories else "Autre")
    
from sklearn.preprocessing import LabelEncoder

label_encoder = LabelEncoder()
for col in ["dep", "com", "adr", "actp"]:
    df_clean[col] = label_encoder.fit_transform(df_clean[col])

df_encoded = pd.get_dummies(df_clean, columns=["dep", "com", "adr", "actp"], drop_first=True, sparse=True)

df_clean.drop(columns=["adr", "actp"], inplace=True) 
df_encoded = pd.get_dummies(df_clean, columns=["dep", "com"], drop_first=True)
```

```python
from sklearn.preprocessing import StandardScaler

scaler = StandardScaler()
df_encoded[["an_nais", "lat", "long", "v1"]] = scaler.fit_transform(df_encoded[["an_nais", "lat", "long", "v1"]])
```

```python
cat_cols = df_encoded.select_dtypes(include=['object']).columns
print("Colonnes à convertir :", cat_cols)
```

```python
for col in ['hrmn', 'voie', 'pr', 'pr1', 'larrout']:
    print(f"Colonne {col} - Nombre de valeurs uniques : {df_encoded[col].nunique()}")

df_encoded.drop(columns=["hrmn", "pr", "pr1"], inplace=True)

from sklearn.preprocessing import LabelEncoder

label_encoder = LabelEncoder()
for col in ["voie", "larrout"]:
    df_encoded[col] = label_encoder.fit_transform(df_encoded[col])

df_encoded = pd.get_dummies(df_encoded, columns=["voie", "larrout"], drop_first=True)
```
