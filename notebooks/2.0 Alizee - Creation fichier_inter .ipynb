```python
import pandas as pd
```

Pour creer le DataFrame final nous allons procéder ainsi :
- Grouper chaque type(caract, lieux,usager) de fichier avec toutes les années
/!\ Il faut isoler année 2005-2018 et 2018-2023
- merger les fichiers ensemble sur colonne "Num_Acc"
/!\Penser a verifier le nombre de colonne
- Exporter le csv

```python
# DF caracteristiques
# Chemin vers les fichiers
data_path = "../data/raw/"

# Listes des années de 2005 à 2016
# /!\ Bien verifier le nom de ces fichiers raw
annees_2005_2016 = list(range(2005, 2017))
df_caracteristiques_2005_2016_list = []

# Fusion an 2005 2016
for annee in annees_2005_2016:
    caracteristiques_file = f"{data_path}/caracteristiques_{annee}.csv"
    separator = ',' if annee != 2009 else '\t'  # Utiliser ',' sauf pour 2009 où on utilise la tabulation
    caracteristiques_df = pd.read_csv(caracteristiques_file, sep=separator, low_memory=False, encoding='latin-1')
    df_caracteristiques_2005_2016_list.append(caracteristiques_df)

df_caracteristiques_2005_2016 = pd.concat(df_caracteristiques_2005_2016_list, ignore_index=True)
# Pareil pour 2017 2018

annees_2017_2018 = list(range(2017, 2019))
df_caracteristiques_2017_2018_list = []

# Fusion an 2019 2023
for annee in annees_2017_2018:
    caracteristiques_file = f"{data_path}/caracteristiques-{annee}.csv"
    caracteristiques_df = pd.read_csv(caracteristiques_file, sep=',', low_memory=False, encoding='latin-1')
    df_caracteristiques_2017_2018_list.append(caracteristiques_df)

df_caracteristiques_2017_2018 = pd.concat(df_caracteristiques_2017_2018_list, ignore_index=True)

# Pareil pour 2019 2023
# Listes des années de 2019 à 2023, ! fichier 2023
annees_2019_2023 = list(range(2019, 2024))
df_caracteristiques_2019_2023_list = []

# Fusion an 2019 2023
for annee in annees_2019_2023:
    caracteristiques_file = f"{data_path}/caracteristiques-{annee}.csv"
    caracteristiques_df = pd.read_csv(caracteristiques_file, sep=';', low_memory=False, encoding='latin-1')
    df_caracteristiques_2019_2023_list.append(caracteristiques_df)

df_caracteristiques_2019_2023 = pd.concat(df_caracteristiques_2019_2023_list, ignore_index=True)

# Concat de 2005 a 2023
df_caracteristiques_final = pd.concat([df_caracteristiques_2005_2016, df_caracteristiques_2017_2018, df_caracteristiques_2019_2023], ignore_index=True)

     
```

```python
unique_carac_final = df_caracteristiques_final['Num_Acc'].nunique()
print("Nombre de valeurs uniques pour 'Num_Acc':", unique_carac_final)
```

```python
import pandas as pd
import os

# Définir les chemins
RAW_PATH = "../data/raw/"
PROCESSED_PATH = "../data/processed/"

### Étape 1 : fusion usagers + véhicules ###
if not os.path.exists(f"{PROCESSED_PATH}df_UV.csv"):
    print("Fusion usagers + véhicules...")
    usagers = pd.read_csv(f"{PROCESSED_PATH}usagers_final.csv")
    vehicules = pd.read_csv(f"{PROCESSED_PATH}vehicules_final.csv")
    
    df_UV = pd.merge(usagers, vehicules, on=["Num_Acc", "num_veh"], how="inner")
    print("UV shape:", df_UV.shape)
    df_UV.to_csv(f"{PROCESSED_PATH}df_UV.csv", index=False)
    del usagers, vehicules, df_UV

### Étape 2 : fusion UV + lieux ###
if not os.path.exists(f"{PROCESSED_PATH}df_UVL.csv"):
    print("Fusion UV + lieux...")
    df_UV = pd.read_csv(f"{PROCESSED_PATH}df_UV.csv")
    lieux = pd.read_csv(f"{PROCESSED_PATH}lieux_final.csv")
    
    df_UVL = pd.merge(df_UV, lieux, on="Num_Acc", how="inner")
    print("UVL shape:", df_UVL.shape)
    df_UVL.to_csv(f"{PROCESSED_PATH}df_UVL.csv", index=False)
    del df_UV, lieux, df_UVL

### Étape 3 : fusion UVL + caractéristiques ###
if not os.path.exists(f"{PROCESSED_PATH}df_final.csv"):
    print("Fusion UVL + caractéristiques...")
    df_UVL = pd.read_csv(f"{PROCESSED_PATH}df_UVL.csv")
    caracteristiques = pd.read_csv(f"{PROCESSED_PATH}caracteristiques_final.csv")
    
    df = pd.merge(df_UVL, caracteristiques, on="Num_Acc", how="inner")
    print("Final df shape:", df.shape)
    df.to_csv(f"{PROCESSED_PATH}df_final.csv", index=False)
    del df_UVL, caracteristiques, df

print("✅ Toutes les fusions ont été effectuées sans plantage.")
```
