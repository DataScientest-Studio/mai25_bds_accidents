```python
import pandas as pd
import numpy as np

accidents_2019_2023 = pd.read_csv("../data/raw/accidents_2019_2023.csv")
accidents_2004_2023 = pd.read_csv("../data/processed/Accident_auto_combined.csv")
```

```python
accidents_2004_2023['an'] = accidents_2004_2023['Num_Acc'].astype(str).str[:4]
```

```python
accidents_2004_2018 = accidents_2004_2023[accidents_2004_2023['an'].astype(int).between(2004, 2018)]
```

```python
accidents = pd.concat([accidents_2019_2023,accidents_2004_2018], axis=0, join='inner', ignore_index=True)
```

```python
accidents.to_csv("../data/raw/accidents_2004_2023.csv", index=False)
```
