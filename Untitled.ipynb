```python
df['age'] = df['annee'] - df['an_nais']
```

```python
# Créer des tranches d'âge
bins = [0, 9, 19, 29, 39, 49, 59, 69, 79, 89, 99, 150]
labels = ['0-9', '10-19', '20-29', '30-39', '40-49',
          '50-59', '60-69', '70-79', '80-89', '90-99', '100+']
df['tranche_age'] = pd.cut(df['age'], bins=bins, labels=labels, right=True)

# Tracer le graphique
plt.figure(figsize=(10, 6))
df['tranche_age'].value_counts().sort_index().plot(kind='bar', color='mediumseagreen')
plt.title("Nombre d'accidents par tranche d'âge", fontsize=14)
plt.xlabel("Tranche d'âge")
plt.ylabel("Nombre d'accidents")
plt.xticks(rotation=45)
plt.show()
```

```python
plt.figure(figsize=(18, 9))

# Sélection des variables numériques
df_num_features = df.select_dtypes(include=np.number)

# Calcul de la matrice de corrélation
corr = df_num_features.corr(method='pearson')

# Affichage de la heatmap
sns.heatmap(corr, annot=True, cmap='tab20c')
plt.title("Matrice de corrélation des variables numériques", fontsize=16)
plt.show()
```

```python
# Corrélations avec la variable cible "grav"
corr_grav = corr['grav'].drop('grav').sort_values(ascending=False)
plt.figure(figsize=(10, 6))
sns.barplot(x=corr_grav.values, y=corr_grav.index, palette='coolwarm')
plt.title("Corrélations avec la gravité de l'accident")
plt.xlabel("Corrélation")
plt.tight_layout()
plt.show()
```

```python
df = df.apply(pd.to_numeric, errors='coerce')
print(df.isna().sum())
```

```python
import statsmodels.api as sm

X = df[["an_nais", "sexe", "place", "choc", "secu1"]]  # Variables explicatives
y = df["grav"]  # Variable cible avec plusieurs niveaux

X = sm.add_constant(X)  # Ajout de la constante
model = sm.MNLogit(y, X).fit()  # Modèle multinomial
print(model.summary())
```

```python
Pseudo R² = 0.04371

# Cela indique que le modèle explique environ 4.37 % de la variance dans la gravité des accidents. C'est un score modéré, ce qui suggère que d'autres facteurs non inclus pourraient influencer les résultats.
#Significativité des variables (P>|z|)

#Toutes les variables ont des p-values très faibles (≈ 0.000), ce qui signifie qu'elles ont un impact statistiquement significatif sur la gravité des accident
#Effet des variables selon le niveau de gravité :
#Année de naissance (an_nais) : un coefficient négatif pour grav=2 et 3 mais positif pour grav=4 → Les accidents graves (niveau 4) semblent plus fréquents chez les plus jeunes.

#Sexe (sexe) : négatif pour grav=2, positif pour grav=3 et 4 → Influence différente selon le niveau de gravité.

#Place dans le véhicule (place) : un coefficient positif, indiquant que l’endroit où la personne est assise a un effet sur la gravité.

#Type de choc (choc) : coefficient toujours négatif, suggérant que certains types de collisions réduisent la gravité.

#Sécurité (secu1) : effet positif, ce qui peut signifier que certains équipements de sécurité influencent la gravité.
```
