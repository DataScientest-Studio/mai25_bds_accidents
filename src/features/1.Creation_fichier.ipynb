```python
import pandas as pd
import numpy as np
import os

# Définir le chemin de base relatif vers les fichiers CSV
data_path = os.path.join("..", "..", "data", "raw")

# Chargement des fichiers
def load_csv(filename):
    return pd.read_csv(os.path.join(data_path, filename), sep=';')

# 2019
usagers_2019 = load_csv("usagers-2019.csv")
caract_2019 = load_csv("caracteristiques-2019.csv")
lieux_2019 = load_csv("lieux-2019.csv")
vehicules_2019 = load_csv("vehicules-2019.csv")

# 2020
usagers_2020 = load_csv("usagers-2020.csv")
caract_2020 = load_csv("caracteristiques-2020.csv")
lieux_2020 = load_csv("lieux-2020.csv")
vehicules_2020 = load_csv("vehicules-2020.csv")

# 2021
usagers_2021 = load_csv("usagers-2021.csv")
caract_2021 = load_csv("caracteristiques-2021.csv")
lieux_2021 = load_csv("lieux-2021.csv")
vehicules_2021 = load_csv("vehicules-2021.csv")

# 2022
usagers_2022 = load_csv("usagers-2022.csv")
caract_2022 = load_csv("caracteristiques-2022.csv")
lieux_2022 = load_csv("lieux-2022.csv")
vehicules_2022 = load_csv("vehicules-2022.csv")

# 2023
usagers_2023 = load_csv("usagers-2023.csv")
caract_2023 = load_csv("caracteristiques-2023.csv")
lieux_2023 = load_csv("lieux-2023.csv")
vehicules_2023 = load_csv("vehicules-2023.csv")

# Harmonisation des noms de colonnes
caract_2022 = caract_2022.rename(columns={'Accident_Id': 'Num_Acc'})

# Fusion des jeux de données
def merge_accidents(usagers, vehicules, caract, lieux):
    veh_usagers = vehicules.merge(usagers, on=['Num_Acc', 'id_vehicule', 'num_veh'], how='inner')
    caract_lieux = caract.merge(lieux, on='Num_Acc')
    return veh_usagers.merge(caract_lieux, on='Num_Acc')

accidents_2019 = merge_accidents(usagers_2019, vehicules_2019, caract_2019, lieux_2019)
accidents_2020 = merge_accidents(usagers_2020, vehicules_2020, caract_2020, lieux_2020)
accidents_2021 = merge_accidents(usagers_2021, vehicules_2021, caract_2021, lieux_2021)
accidents_2022 = merge_accidents(usagers_2022, vehicules_2022, caract_2022, lieux_2022)
accidents_2023 = merge_accidents(usagers_2023, vehicules_2023, caract_2023, lieux_2023)

# Concaténation de toutes les années
accidents_copy= pd.concat(
    [accidents_2019, accidents_2020, accidents_2021, accidents_2022, accidents_2023],
    axis=0,
    join='inner',
    ignore_index=True
)

# Sauvegarde dans le dossier raw
output_path = os.path.join(data_path, "accidents_2019_2023.csv")
accidents_copy.to_csv(output_path, index=False)
```
